```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import minimum_filter, maximum_filter
```

```python
image_path = "images_1.jpg"
image = cv2.imread(image_path)

image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
median_filtered = cv2.medianBlur(image_rgb,15)
```

```python
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(image_rgb)
plt.axis('off')
plt.title("Original Image")
plt.subplot(1, 2, 2)
plt.imshow(median_filtered)
plt.axis('off')
plt.title("Median Filtered Image")
plt.show()
```

```python
bilateral_filtered = cv2.bilateralFilter(image_rgb, 200, 175, 175)
```

```python
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(image_rgb)
plt.axis('off')
plt.title("Original Image")
plt.subplot(1, 2, 2)
plt.imshow(bilateral_filtered)
plt.axis('off')
plt.title("Bilateral Filtered Image")
plt.show()
```

```python
minimum_filtered = minimum_filter(image_rgb, size=5)
maximum_filtered = maximum_filter(image_rgb, size=5)
```

```python
plt.figure(figsize=(10, 5))
plt.subplot(1, 3, 1)
plt.imshow(image_rgb)
plt.axis('off')
plt.title("Original Image")
plt.subplot(1, 3, 2)
plt.imshow(minimum_filtered)
plt.axis('off')
plt.title("Minimum Filtered Image")
plt.subplot(1, 3, 3)
plt.imshow(maximum_filtered)
plt.axis('off')
plt.title("Maximum Filtered Image")
plt.show()
```
