```python
import numpy as np
import matplotlib.pyplot as plt

signal = np.array([1, 2, 3, 4, 5, 6, 7, 8])
n = np.arange(len(signal))

# Calculate FFT, magnitude and phase.
fft_result = np.fft.fft(signal)
magnitude = np.abs(fft_result)
phase = np.angle(fft_result)

# Recover the original signal using IFFT.
reconstructed = np.fft.ifft(fft_result)

# Plot the input, spectra and reconstructed signal.
data = [signal, magnitude, phase, reconstructed.real]
titles = ['Original signal', 'FFT magnitude spectrum',
          'FFT phase spectrum', 'Reconstructed signal using IFFT']
labels = ['Amplitude', 'Magnitude', 'Phase (radians)', 'Amplitude']

plt.figure(figsize=(10, 7))
for i in range(4):
    plt.subplot(2, 2, i + 1)
    plt.stem(n, data[i])
    plt.title(titles[i])
    plt.xlabel('Frequency bin k' if i in [1, 2] else 'Sample n')
    plt.ylabel(labels[i])
    plt.xticks(n)
    plt.grid(True)
plt.tight_layout()
plt.show()

print('FFT:', np.round(fft_result, 4))
print('IFFT:', np.round(reconstructed.real, 4))
print('Reconstruction correct:', np.allclose(signal, reconstructed))
```

**Conclusion:** FFT converted the 8-sample signal into the frequency domain. IFFT recovered `[1, 2, 3, 4, 5, 6, 7, 8]` within floating-point precision.
